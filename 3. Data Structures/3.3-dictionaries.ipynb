## Dictionary

- unordered collection of items
- key-value pairs
- keys unique and immutable (string, numbers, tuples)
- values can be any type

```python
# creating dictionary

empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={'name':'suryansh', 'age':23, "grade":"A"}
print(student)
```

```python
# error
student = {"name": "suryansh", "age": 23, "name": 12} #replaces the value
print(student)
```

```python
# accessing
student = {"name": "suryansh", "age": 23, "grade": 'A'}
student["grade"]
student["age"]
```

```python
student.get('grade')
```

```python
student.get('last_name')
```

```python
student.get('last_name', 'Not Available')
```

```python
# Modifying dict elements

## dictionary are MUTABLE, can add, update or delete elems
print(student)
```

```python
student['age']=33 #udpate value
print(student)
student['address']='singapore' #add a new key value
print(student)
```

```python
del student['grade']
print(student)
```

```python
# dictionary methods

keys=student.keys()
print(keys)
values=student.values()
print(values)
```

```python
items=student.items()
print(items)

# this is a list of tuples as output
```

```python
# pointers
student_copy=student
print(student)
print(student_copy)
```

```python
student['name']='kushwaha'
print(student)
print(student_copy)
```

```python
# shallow copy

student_copy1=student.copy() 
print(student)
print(student_copy1)
```

```python
student['name']='surry'
print(student)
print(student_copy1)
```

```python
# iteration

## dict got key, value or items(the key, value pair)

# iterating over keys
for key in student.keys():
    print(key)
```

```python
# iterating over values
for value in student.values():
    print(value)
```

```python
for key, value in student.items():
    print(key, value)
```

```python
# Nested dictionaries

students={
    'student1': {'name':'suryansh','age':23},
    'student2': {'name':'peter','age':35}
}
```

```python
students
```

```python
print(students['student2']['name'])
```

```python
# iterating over nested dict
for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
```

```python
# dict comprehension

squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## conditional dict comprehension

evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
# PRACTICAL EXAMPLES

numbers=[1,2,2,3,3,3,4,4,4,4]

freq_dict=dict()

for item in numbers:
    freq_dict[item]=freq_dict.get(item,0)+1

print(freq_dict)
```

```python
# merge 2 dicts into one

dict1={'a':1,'b':2}
dict2 = {"d": 3, "c": 4}

merged_dict={**dict1, **dict2}
print(merged_dict)
```
