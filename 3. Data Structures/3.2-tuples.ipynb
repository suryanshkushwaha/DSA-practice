### Tuples

Ordered collection of items
immutable

```python
# create a tuple

empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
# list to tuple

nums=tuple([1,2,3,4,5,6,7])
nums
```

```python
# tuple to list

list((1,2,3,4,5,6))
```

```python
mixed_tpl=(1, 'hello world', 3.14, True)

print(mixed_tpl)
```

```python
# accessing tuple elements

nums
```

```python
print(nums[2])
print(nums[0:4])
print(nums[::2])
```

```python
# tuple operations

nums+mixed_tpl
```

```python
mixed_tpl*3
```

```python
# immutability

lst=[1,2,3,4,5]
print(lst)

lst[1]='suryansh'
print(lst)
```

```python
nums[1]='suryansh'
```

```python
# tuple methods

nums.count(1)
```

```python
nums.index(3)
```

```python
# packing & unpacking

# packing
packed_tpl=1, 'hello', 3.14
print(packed_tpl)
```

```python
# unpacking
a,b,c=packed_tpl

print(a)
print(b)
print(c)
```

```python
# unpacking with *

first, *middle, last = nums
print(first)
print(middle) # notice how it returned a list
print(last)
```

```python
# nested list
lst=[[1,2,3,4],['hi','waasup', 'perf']]
lst[0][0:3]

# nested tuple
```

```python
lst = [[1, 2, 3, 4], ("hi", "waasup", "perf")]
lst[1][0:2] # notice it returns a tuple
```

```python
nested_tpl=((1,2,3),('a','b','c'),(True, False))
```

```python
# iterating over nested tuples
for sub_tpl in nested_tpl:
    for item in sub_tpl:
        print(item, end=" ")
    print()
```
