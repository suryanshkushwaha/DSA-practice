#### Sets
- another build in data type
- collections of UNIQUE ITEMS - no duplicates allowed
- unordered

##### uses
- membership tests
- eliminate duplicates
- maths operations like union, intersection, difference, symmetric

```python
# create a set

my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(type(my_empty_set))
```

```python
my_new_set = set([1, 2, 3, 4, 5])
print(my_new_set)
```

```python
no_duplicates_set = {1, 1, 1, 4, 3, 2, 3, 4, 5, 6, 6, 3}
print(no_duplicates_set)
```

```python
# set membership test

st = {1, 2, 3, 4, 5}
print(3 in st)
print(10 in st)
```

```python
# mathematical operations

set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

## Union
union_set = set1.union(set2)
print(union_set)
```

```python
## intersection
intersect_set = set1.intersection(set2)
print(intersect_set)
```

```python
set1.intersection_update(set2)  # udpates set 1 with the intersection
```

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

## difference

print(set1.difference(set2))
```

```python
## symmetric difference

set1.symmetric_difference(set2)
```

```python
## Set Methods

set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

print(set1.issubset(set2))
```

```python
set1.issuperset(set2)
```
