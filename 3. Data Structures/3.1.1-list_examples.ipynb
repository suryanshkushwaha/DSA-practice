Example1. Manage a todo list

create todo list to keep track of things

```python
to_do_list=['buy groceries','clean the house', 'pay the bills']

## add to the task

to_do_list.append('shampoo dog')
to_do_list.append('go for a run')

## remove the task
to_do_list.remove('clean the house')

## check if task in the list
if 'pay the bills' in to_do_list:
    print('dont forget to pay the bills')

print('tasks remanining', to_do_list)
```

Example2. Organizing student grades

createa a list to store and calculate average grades for students

```python
grades = [10, 23,45,24,99]

grades.append(100)

average_grade=sum(grades)/len(grades)
max_grade=max(grades)
min_grade=min(grades)

print('avg: ', average_grade)
print("max: ", max_grade)
print("min: ", min_grade)
```
