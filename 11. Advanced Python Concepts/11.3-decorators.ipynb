# Decorators

- powerful and flexible feature
- allows to modify the behavior of function or class method
- commonly used to add functionality to functions or methods without modifying thier actual code

```python
## function copy
## closures
## decorators
```

```python
# function copy


def welcome():
    return "welcome to singapore"


welcome()
```

```python
wel = welcome

wel()
```

```python
del welcome
```

```python
wel()

# still present even after deletion
```

```python
# CLOSURES


def main_welcome():
    msg = "welcome"

    def sub_welcome():
        print("welcome to singapore lah")
        print(f"you can access variables defined outside --> {msg}")
        print("please do well in singapore")

    return sub_welcome()
```

```python
main_welcome()
```

```python
# CLOSURES


def main_welcome(func):
    def sub_welcome():
        print("welcome to singapore lah")
        func("welcome to printing")
        print("please do well in singapore")

    return sub_welcome()
```

```python
main_welcome(print)
```

```python
def main_welcome(func, lst):
    def sub_welcome():
        print(func(lst))
        print("please do well in singapore")

    return sub_welcome()
```

```python
main_welcome(len, [1, 2, 3, 4])
```

```python
def main_welcome(func):
    def sub_welcome():
        print("welcome to singapore lah")
        func()
        print("please do well in singapore")

    return sub_welcome()
```

```python
def learning_python():
    print("i am learning python")


learning_python()
```

```python
main_welcome(learning_python)
```

```python
@main_welcome
def learning_python():
    print("i am learning python")


# now this function acts like a parameter to the main_welcome function
```

```python
def my_decorator(func):
    def wrapper():
        print("before the func call")
        func()
        print("after func call")

    return wrapper
```

```python
@my_decorator
def say_hello():
    print("hello")
```

```python
say_hello()
```

```python
# decorator with arguments


def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)

        return wrapper

    return decorator
```

```python
@repeat(3)
def say_hello():
    print("hello")
```

```python
say_hello()
```
