# Generators

- subclass of iterators
- simpler way to create iterators
- yield keyword to produce series of values lazily
- generate values on the FLY
- do not store them in memory

```python
def square(n):
    for i in range(3):
        return i**2


square(3)
```

```python
def square(n):
    for i in range(3):
        yield i**2


square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen = my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

# PRACTICAL EXAMPLE - READING LARGE FILE

- allows to process one line at a time without loading entire file in memory

```python
def read_large_file(file_path):
    with open(file_path, "r") as file:
        for line in file:
            yield line
```

```python
line = read_large_file("./large_file.txt")
```

```python
next(line)
```

```python
for line in read_large_file("./large_file.txt"):
    print(line.strip())
```
