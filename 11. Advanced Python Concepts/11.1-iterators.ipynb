# Iterators

- allow for efficient looping and memory management.
- allow to access elements of collection sequentially without exposing the underlying structure

```python
my_lst = [1, 2, 3, 4, 5, 6]

for i in my_lst:
    print(i)
```

```python
iterator = iter(my_lst)
print(type(iterator))
```

```python
my_lst = [1, 2, 3, 4, 5, 6]
iterator = iter(my_lst)

try:
    print(next(iterator))
except StopIteration:
    print("There are no more elements")
```
