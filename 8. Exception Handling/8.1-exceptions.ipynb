# Exceptions

- handle erros gracefully
- take corrective actions, WITHOUT stopping the execution of the program

### Common exceptions

- ZeroDivisionError: diving by zero
- FileNotFoundError: file not found
- ValueError: invalid value
- TypeError: invalid type

```python
a = b
```

```python
# try, except block

try:
    a = b
except NameError as ne:
    print(ne)
```

```python
result = 1 / 0
```

```python
try:
    result = 1 / 0
except ZeroDivisionError as ex:
    print(ex)
    print("please enter denominator greator than 0")
```

```python
# unable to catch all errors

try:
    result = 1 / 2
    a = b
except ZeroDivisionError as ex:
    print(ex)
    print("please enter denominator greator than 0")
```

```python
try:
    result = 1 / 2
    a = b
except ZeroDivisionError as ex:
    print(ex)
    print("please enter denominator greator than 0")
except Exception as ex:
    print(ex)
```

```python
try:
    num = int(input("enter a no"))
    result = 10 / num
except ValueError:
    print("not a valid no")
except ZeroDivisionError:
    print("enter denominator greator than 0")
except Exception as ex:
    print(ex)
```

```python
# try, except, else block

try:
    num = int(input("enter a no"))
    result = 10 / num
except ValueError:
    print("not a valid no")
except ZeroDivisionError:
    print("enter denominator greator than 0")
except Exception as ex:
    print(ex)
else:
    print(f"the result is {result}")
```

```python
# try, except, else, finally

try:
    num = int(input("enter a no"))
    result = 10 / num
except ValueError:
    print("not a valid no")
except ZeroDivisionError:
    print("enter denominator greator than 0")
except Exception as ex:
    print(ex)
else:
    print(f"the result is {result}")
finally:
    print("execution complete")  # will get executed even in case of error or not
```
