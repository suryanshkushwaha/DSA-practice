# Trees

- non-linear heirarchial data
- data structure consisting of nodes which are connected by edges

### Examples
- family tree
- organizational chart

### Things to Note
- can be as many children as you want
- there should be no cycle
- only 1:1 relation
- tree is either None or a tree

### Practical Examples
- file structure
- decision trees

### Industry Examples
- indexing in db
- network routing
- HTML DOM


# Terminology

1. Node - each item of tree
2. Edge - connection b/w two nodes
3. Root - Topmost node
4. Leave - node with no children!
5. Parent - only the next upper node
6. Child - only the next branching node
7. Siblings - children belonging to same parent
8. Cousins - children belonging to same ancestor at the same level
9. Ancestors - all the connected nodes above (Root node is ancestor of the entire tree)
10. Descendants - all the connected nodes below
11. Subtree - group of nodes which creates a tree within the parent tree
12. Level / Depth - how many edges from the root node (starting from 0)
13. Path - edges taken from one node to another


# Tree Implementation


```python
"""
# Node
1. Data
2. Children
    - List
        - indexing
        - dynamic nature
    - LinkedList
        - head and tail
        - hard to traverse
"""


class TreeNode:
    def __init__(self, data) -> None:
        self.data = data
        self.children = []  # picked list since better indexing


root = TreeNode(1)

child1 = TreeNode(2)
child2 = TreeNode(3)
child3 = TreeNode(4)

root.children.append(child1)
root.children.append(child2)
root.children.append(child3)

print(root.children[0].data)
print(root.children[1].data)
print(root.children[2].data)
```

### Print Tree

```python
def print_tree(root):
    print(root.data)

    for child in root.children:
        print_tree(child)


print_tree(root)

# print_tree(None)
# this would give error must handle this edge case

# problem here is that u can differentiate between
# Root 1 with children 2, 3, 4
# VS
# Root 1 with children (Root 2 with child 3) and 4
# both gives the same result
```

```python
def print_tree_detailed(root):
    if root is None:
        return

    print(root.data, end=": ")

    print(*[child.data for child in root.children], sep=", ")

    for child in root.children:
        print_tree_detailed(child)


print_tree_detailed(root)
```

### Take Input

```python
def take_input():
    value = int(input("Enter the node value"))

    root = TreeNode(value)

    num_children = int(input(f"Enter no of children for the node: {value}"))

    for _ in range(num_children):
        root.children.append(take_input())

    return root


tree = take_input()
print_tree_detailed(tree)
```

```python
from collections import deque


def take_input_levelwise():
    parent_root = int(input("Enter node value"))
    parent_root = TreeNode(parent_root)

    q = deque([parent_root])

    while q:
        curr = q.popleft()
        num_children = int(input(f"Enter no of children for {curr.data}: "))

        for i in range(num_children):
            child_data = int(
                input(f"Enter the data for {i+1} child for parent {curr.data}: ")
            )
            child_node = TreeNode(child_data)
            curr.children.append(child_node)
            q.append(child_node)

    return parent_root


tree = take_input_levelwise()
print_tree_detailed(tree)
```

```python
def count_nodes(tree):
    if tree is None:
        return 0

    count = 1
    for child in tree.children:
        count += count_nodes(child)

    return count


count_nodes(tree)
```

```python
tree = take_input_levelwise()
```

```python
def height(tree):
    if tree is None:
        return 0

    heighest = 0

    for child in tree.children:
        child_ht = height(child)
        heighest = max(heighest, child_ht)

    return heighest + 1


print_tree_detailed(tree)
height(tree)
```

# Pre/Post Order Traversal


----------------


## Pre Order: First Parent --> then Child

---------

## Post Order: First Child --> then Parent

```python
def pre_order_traverse(tree):
    if tree is None:
        return

    print(tree.data, end=" ")

    for child in tree.children:
        pre_order_traverse(child)


pre_order_traverse(tree)
```

```python
def post_order_traverse(tree):
    if tree is None:
        return

    for child in tree.children:
        post_order_traverse(child)

    print(tree.data, end=" ")


post_order_traverse(tree)
```
