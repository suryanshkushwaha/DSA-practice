# Queue

- Queue is a queue
- Linear Data Structure
- FIFO: first in first out

#### Real Life example
- printer jobs
- customer support
- elevator request
- task scheduling

#### Industry based examples
- job scheduling
- network traffic
- messaging service (eg. kafka, rabbit mq)
- BFS

```python
"""
            ------------
            |           |
[Rear]in -->   contents   --> out [Front]
            |           |
            ------------
"""
```

### Operations

- enqueue: add elem to the end of queue
- dequeue: remove elem from front of queue
- size: lenght of queue
- isEmpty: true or false
- front: elem at the front of queue [Note: this is the oldest element, it was the first to enter and will get removed first also. But this is the oldest amongst all other elems]
- rear: last elem to enter queue [Note: this is the newest elem.]

# Queue using LISTS

```python
"""
List = [1, 2, 3, 4, 5]

             --------------
             |             |
[Rear] in --> 5, 4, 3, 2, 1  --> out [Front]
             |             |
             --------------


this is how we will be implementing out q using list
"""
```

```python
class QueueUsingList:
    def __init__(self) -> None:
        self.__queue = []

    def enqueue(self, item):
        self.__queue.append(item)

    def size(self):
        return len(self.__queue)

    def is_empty(self):
        return self.size() == 0

    def front(self):
        return "empty queue" if self.is_empty() else self.__queue[0]

    def dequeue(self):
        return None if self.is_empty() else self.__queue.pop(0)
```

```python
q = QueueUsingList()
print(q.is_empty())  # True

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print(q.front())  # 10
print(q.size())  # 3

print(q.dequeue())  # 10
print(q.front())  # 20
print(q.size())  # 2
```

# Queue using LINKEDLIST

### Head -> 1 -> 2 -> 3 -> 4 -> None [Insert Node at tail]

- enqueue: possible in O(1)
    - maintain `tail`
- front: O(1)
    - head.data
- dequeue: O(1)
    - head = head.next
- size: O(1)
    - use `size` variable    
    - traverse through entire LL if no size then O(n)
- is_empty: O(1)

```python
class Node:
    def __init__(self, data) -> None:
        self.data = data
        self.next = None


class QueueUsingLinkedList:
    def __init__(self) -> None:
        self.__head = None
        self.__length = 0
        self.__tail = None

    def size(self):
        return self.__length

    def is_empty(self):
        return self.size() == 0

    def front(self):
        return "empty queue" if self.is_empty() else self.__head.data

    def enqueue(self, item):
        newNode = Node(item)
        if self.__head is None:
            self.__head = newNode
            self.__tail = newNode
        else:
            self.__tail.next = newNode
            self.__tail = self.__tail.next
        self.__length += 1

    def dequeue(self):
        if self.is_empty():
            return None
        else:
            item = self.__head.data
            self.__head = self.__head.next
            if self.__head is None:  ## VERY IMPORTANT: THINK WHY
                self.__tail = None
            self.__length -= 1
            return item
```

```python
q = QueueUsingLinkedList()
print(q.is_empty())  # True

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print(q.front())  # 10
print(q.size())  # 3

print(q.dequeue())  # 10
print(q.front())  # 20
print(q.size())  # 2
```

## Types of Queues

### Simple Queue
- LIFO
- single entry and single exit
- which i js implemented above

### Circular Queue
- similar to circular LL
- eg: days in a week 
- last node is connected to first node
- need to remember where to started from (or rmb the first node)
- ALSO KNOWN AS **RING BUFFER**

### Priority Queue
- each element has a priority attached to it
- eg: giving priority to people who had already made a reservation
- elements are served/popped based on priority


### Double Ended Queue = DeQueue
- double entry and exit points
- can enter from the start or end 
- can exit from start or end
- `from collections import deque`
