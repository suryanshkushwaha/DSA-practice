# Binary Search Tree (BST)


- each node of tree follows 
     - left subtree contains node whose value is less than root's value
     - right subtree contains node whose value is greater than root's value
     - left and right subtree must also be BST

- inspired by Binary Search Algorithm [O(n) -> O(log n)]

```python
# Example BST structure:
#           50
#         /    \
#       30      70
#      /  \    /  \
#    20   40  60   80
#
# Inorder traversal (sorted): 20, 30, 40, 50, 60, 70, 80
# BST property:
# - left child  < parent
# - right child > parent
```

## Real World Example
1. Phone book / Dictionary
2. File Searching
3. Autocomplete 

## Industry Example
1. Database (indexing)
2. ML (Decision Tree)
3. Range Queries in DB


```python
"""
the structure is same as BT node
"""


class BSTNode:
    def __init__(self, data) -> None:
        self.data = data
        self.left = None
        self.right = None


def print_BST(root):
    if root is None:
        return

    print_BST(root.left)
    print(root.data, end=" ")
    print_BST(root.right)
```

```python
# Examples

# 1) Balanced BST
#        4
#      /   \
#     2     6
#    / \   / \
#   1  3  5   7
root1 = BSTNode(4)
root1.left = BSTNode(2)
root1.right = BSTNode(6)
root1.left.left = BSTNode(1)
root1.left.right = BSTNode(3)
root1.right.left = BSTNode(5)
root1.right.right = BSTNode(7)

# 2) Skewed-left BST
#      4
#     /
#    3
#   /
#  2
# /
# 1
root2 = BSTNode(4)
root2.left = BSTNode(3)
root2.left.left = BSTNode(2)
root2.left.left.left = BSTNode(1)

# 3) Skewed-right BST
# 1
#  \
#   2
#    \
#     3
#      \
#       4
root3 = BSTNode(1)
root3.right = BSTNode(2)
root3.right.right = BSTNode(3)
root3.right.right.right = BSTNode(4)

# 4) Unbalanced BST
#      5
#     / \
#    3   8
#   /   / \
#  2   6   9
#       \
#        7
root4 = BSTNode(5)
root4.left = BSTNode(3)
root4.left.left = BSTNode(2)
root4.right = BSTNode(8)
root4.right.left = BSTNode(6)
root4.right.left.right = BSTNode(7)
root4.right.right = BSTNode(9)
```

```python
def search(root, value):
    if root is None:
        return False

    if value == root.data:
        return True
    elif value < root.data:
        return search(root.left, value)
    else:
        return search(root.right, value)


search(root3, 4)
```

```python
def sorted_list_to_BST(lst):
    if len(lst) == 0:
        return None

    mid = len(lst) // 2

    rootNode = BSTNode(lst[mid])

    rootNode.left = sorted_list_to_BST(lst[:mid])
    rootNode.right = sorted_list_to_BST(lst[mid + 1 :])

    return rootNode


root = sorted_list_to_BST([1, 2, 3, 4, 5, 6, 7])
print_BST(root)
```

```python
def _checkBST_minmax(tree):
    # returns (is_bst, min_value, max_value)
    if tree is None:
        return True, float("inf"), float("-inf")

    left_is_bst, left_min, left_max = _checkBST_minmax(tree.left)
    right_is_bst, right_min, right_max = _checkBST_minmax(tree.right)

    if not left_is_bst or not right_is_bst:
        return False, None, None

    if not (left_max < tree.data < right_min):
        return False, None, None

    min_value = left_min if tree.left is not None else tree.data
    max_value = right_max if tree.right is not None else tree.data

    return True, min(min_value, tree.data), max(max_value, tree.data)


def checkBST(tree):
    is_bst, _, _ = _checkBST_minmax(tree)
    return is_bst


"""
This version with min/max boolean return type ensures:
- each subtree reports its own min/max boundaries.
- current node is valid if left_max < node < right_min.
- returns True/False from wrapper `checkBST(tree)`.
"""
```

```python
def print_elems_range(root, range):
    if root is None:
        # print(root) # Do this if you want to know if left or right node was null
        return

    data = root.data
    low = range[0]
    high = range[1]

    if low < data:
        print_elems_range(root.left, range)

    if low <= data <= high:
        print(root.data, end=" ")

    if high > data:
        print_elems_range(root.right, range)


print_elems_range(sorted_list_to_BST([10, 30, 40, 50, 60, 70, 80]), [55, 75])
```

```python
def check_BST_using_range(tree, low=float("-inf"), high=float("inf")):
    if tree is None:
        return True
    if not (low < tree.data < high):
        return False
    return checkBST(tree.left, low, tree.data) and checkBST(tree.right, tree.data, high)
```

# BST Class

```python
class BST:
    def __init__(self):
        self.root = None

    def _searchHelper(self, root, value):
        if root is None:
            return False

        if root.data == value:
            return True

        if value < root.data:
            return self._searchHelper(root.left, value)
        else:
            return self._searchHelper(root.right, value)

    def search(self, value):
        return self._searchHelper(self.root, value)

    def _insertHelper(self, root, value):
        newNode = BSTNode(value)
        if root is None:
            root = newNode

        if value < root.data:
            root.left = self._insertHelper(root.left, value)

        if value > root.data:
            root.right = self._insertHelper(root.right, value)

        return root

    def insert(self, value):
        self.root = self._insertHelper(self.root, value)

    def _getMinValue(self, root):
        if root is None:
            return None

        while root:
            minValue = root.data
            root = root.left

        return minValue

    def _deleteHelper(self, root, value):
        if root is None:
            return None

        currNodeValue = root.data

        if value < currNodeValue:
            root.left = self._deleteHelper(root.left, value)

        if value > currNodeValue:
            root.right = self._deleteHelper(root.right, value)

        if value == currNodeValue:
            if not root.left and not root.right:
                return None

            if not root.left:
                return root.right

            if not root.right:
                return root.left

            smallestNextLargerNode = self._getMinValue(root.right)
            root.data = smallestNextLargerNode
            root.right = self._deleteHelper(root.right, smallestNextLargerNode)

        return root

    def delete(self, value):
        self.root = self._deleteHelper(self.root, value)
        return
```
