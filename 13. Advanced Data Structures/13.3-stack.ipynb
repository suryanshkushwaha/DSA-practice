# Stack USING LIST

```python
class StackUsingList:
    def __init__(self) -> None:
        self.stack = []

    def push(self, data):
        self.stack.append(data)

    def size(self):
        return len(self.stack)

    def is_empty(self):
        """
        if len(self.stack) == 0:
            return True
        return False
        """
        return len(self.stack) == 0

    def peek(self):
        # self.stack[self.size()-1]
        return "empty stack" if self.is_empty() else self.stack[-1]

    def pop(self):
        return "empty stack" if self.is_empty() else self.stack.pop()
```

```python
myStack = StackUsingList()
print(myStack.is_empty())
print(myStack.push(1))
print(myStack.push(2))
print(myStack.push(3))
print(myStack.push(4))
print(myStack.push(5))
print(myStack.is_empty())
print(myStack.pop())
print(myStack.pop())
print(myStack.size())
print(myStack.peek())
```

```python
class StackUsingList:
    def __init__(self) -> None:
        self.__stack = []  # Very important to make this private!

    def push(self, data):
        self.__stack.append(data)

    def size(self):
        return len(self.__stack)

    def is_empty(self):
        """
        if len(self.__stack) == 0:
            return True
        return False
        """
        return len(self.__stack) == 0

    def peek(self):
        # self.__stack[self.size()-1]
        return "empty stack" if self.is_empty() else self.__stack[-1]

    def pop(self):
        return "empty stack" if self.is_empty() else self.__stack.pop()
```

### Head -> 5 -> 4 -> 3 -> 2 -> 1 -> None [Insert Node at Head ]
- push() - O(1)
    - directly add to head
- size() - O(n)
    - need to traverse
- is_empty() - O(1)
- peek() - O(1)
    - check head value

    - drop head and point to next

size() is creating an issue. WHY NOT SToRE SIZE AS WELL????

### perfect. Final

- push() - O(1)
    - directly add to head
- size() - O(n)
    - no need to traverse, just return self.size!
- is_empty() - O(1)
- peek() - O(1)
    - check head value
- pop() - O(1)
    - drop head and point to next

```python
class Node:
    def __init__(self, val) -> None:
        self.data = val
        self.next = None


class StackUsingLinkedList:
    def __init__(self):
        self.__stack = None
        self.__size = 0

    def push(self, val):
        newNode = Node(val)
        newNode.next = self.__stack
        self.__size += 1
        self.__stack = newNode
        return newNode

    def size(self):
        return self.__size

    def is_empty(self):
        return self.__size == 0

    def peek(self):
        if self.is_empty():
            return "empty stack"
        return self.__stack.data

    def pop(self):
        if self.is_empty():
            return "empty stack"
        ans = self.peek()
        self.__stack = self.__stack.next
        self.__size -= 1
        return ans
```

```python
myStack = StackUsingLinkedList()
print(myStack.is_empty())
print(myStack.push(1))
print(myStack.push(2))
print(myStack.push(3))
print(myStack.push(4))
print(myStack.push(5))
print(myStack.is_empty())
print(myStack.pop())
print(myStack.pop())
print(myStack.size())
print(myStack.peek())
```
