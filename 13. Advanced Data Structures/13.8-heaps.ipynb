- Heap push: O(log n) - insert
- Heap peek: O(1)

# Max Heap

- max elem always in the node, than it's children
- if you pop, pick the bigger of the children and bubble up
- if you insert, then check if it's at the correct pos, if it's larger than parent, then bubble up

````python
A   = [2, 5, 0, 1, 10, 8, -4, 3, 12, 9]

import heapq
heapq.heapify(A)
A
````

````python
# Heap Push
# O(log n)

heapq.heappush(A, 4)
A
````

````python
# Heap Pop
# O(log n)

minn = heapq.heappop(A)

A, minn
````

````python
# Heap Sort
# Time: # O(n log n)

def heapsort(arr):
    heapq.heapify(arr)

    n = len(arr)
    new_list = [0]*n

    for i in range(n):
        minn = heapq.heappop(arr)
        new_list[i] = minn

    return new_list

heapsort([3,5,2,7,9,1,0,-1])
````

````python
# Heap Push Pop
# O(log n)

heapq.heappushpop(A, 99)
A
````

````python
# Max Heap

A= [3,5,2,7,9,1,0,-1]
n = len(A)

for i in range(n):
    A[i] = -A[i]

heapq.heapify(A)

A
````

````python
'''
Notice how all the max elements are the first
But now accessing becomes a bit tricky
'''

largest = - heapq.heappop(A)
largest

# Similarly for all the operations now you will need to do negation
````

````python
elementToInsert = 99
heapq.heappush(A, -elementToInsert)

A
````

````python
# Heapify using Heap Push

arr = [3,5,2,7,9,1,0,-1]
heap = []

for x in arr:
    heapq.heappush(heap, x)

heap
````
