```python
def take_input():
    value = int(input("Enter value:- "))
    head = None  # always point to the head
    tail = None  # always point to the tail

    while value != -1:
        newNode = Node(value)
        if head is None:
            head = newNode
            tail = newNode
        else:
            tail.next = newNode
            tail = newNode
            # or tail = tail.next bo4th are the same

        value = int(input("Enter value:- "))

    return head


head = take_input()
print_LL(head)
```

```python
def length_LL(head):
    temp = head
    c = 0
    while temp is not None:
        c += 1
        temp = temp.next
    return c


length_LL(head)
```

```python
def length_LL_recursive(head, c):
    if head is None:
        return c
    return length_LL_recursive(head.next, c + 1)


length_LL_recursive(head, 0)
```

```python
def delete_value(head, value):
    if head is None:
        return None

    if head.data == value:
        return head.next

    prev = head

    while prev.next is not None:
        if prev.next.data is value:
            prev.next = prev.next.next
            return head
        prev = prev.next

    print("value not found")
    return head


ll = delete_value(ll, 40)
print_LL(ll)
```

## Array vs LinkedList

| Feature | Array | LinkedList |
|--------|-------|------------|
| Access | O(1) | O(n) |
| Memory management | Fixed size; may have unused slots; no internal fragmentation | Dynamic size; each node stores data + pointer; extra pointer memory; possible external fragmentation |
| Insert at head | O(n) (shift elements) | O(1) |
| Insert at tail | O(1) amortized if capacity available, O(n) if resize needed | O(1) with tail pointer, otherwise O(n) |
| Insert in middle | O(n) (shift elements) | O(n) (traverse + link updates) |
| Best for | Random access, read-heavy workloads | Frequent inserts/deletes (esp. at head), sequential traversal |

### Final LinkedList Class

```python
class Node:
    def __init__(self, val):
        self.data = val
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insertAtHead(self, value):
        newNode = Node(value)
        newNode.next = self.head
        self.head = newNode
        return self.head

    def insertAtIndex(self, value, index):
        if index == 0:
            return self.insertAtHead(value)

        newNode = Node(value)
        prev = self.head
        for _ in range(index - 1):
            if prev is None:
                return "out of bounds"
            prev = prev.next

        if prev is None:
            return "out of bounds"

        newNode.next = prev.next
        prev.next = newNode

        return self.head

    def updateNode(self, valtoUpdate, index):
        if self.head is None:
            return "empty list"

        temp = self.head
        for i in range(index):
            if temp is None:
                return "out of bounds"
            temp = temp.next

        if temp is None:
            return "out of bounds"

        temp.val = valtoUpdate

        return self.head

    def printLL(self):
        temp = self.head

        while temp is not None:
            print(temp.val)
            temp = temp.next

    def lengthOfLL(self):
        temp = self.head
        length = 0
        while temp is not None:
            length += 1
            temp = temp.next

        return length
```

```python
def merge_sorted_ll(head1, head2):
    if head1 is None:
        return head2
    if head2 is None:
        return head1

    finalHead = None
    finalTail = None

    while head1 and head2:
        if head1.data < head2.data:
            if finalHead == None:
                finalHead = head1
                finalTail = head1
            else:
                finalTail.next = head1
                finalTail = head1
            head1 = head1.next

        else:
            if finalHead == None:
                finalHead = head2
                finalTail = head2
            else:
                finalTail.next = head2
                finalTail = head2
            head2 = head2.next

    if head1:
        finalTail.next = head1
    if head2:
        finalTail.next = head2

    return finalHead


ll1 = take_input([2, 3, 9, 10, 17])
# print_LL(ll1)
ll2 = take_input([3, 6, 7])
# print_LL(ll2)
answer = merge_sorted_ll(ll1, ll2)
print_LL(answer)
```

```python
def merge_sorted_ll_recursion(head1, head2):
    if head1 is None:
        return head2
    if head2 is None:
        return head1

    if head1.data < head2.data:
        head1.next = merge_sorted_ll_recursion(head1.next, head2)
        return head1
    else:
        head2.next = merge_sorted_ll_recursion(head1, head2.next)
        return head2


ll1 = take_input([1, 2, 4, 5, 8])
# print_LL(ll1)
ll2 = take_input([1, 3, 4])
# print_LL(ll2)
answer = merge_sorted_ll(ll2, ll1)
print_LL(answer)
```

```python
def reverse_ll_recursion(head):
    if head is None or head.next is None:
        return head

    answer = reverse_ll_recursion(head.next)

    temp = answer
    while temp.next:
        temp = temp.next

    # temp.next = Node(head.data) OR
    temp.next = head
    head.next = None

    return answer


ll1 = take_input([1, 2, 4, 5, 8])
reverse = reverse_ll_recursion(ll1)
print_LL(reverse)

# This is O(n^2)
```

```python
def merge(left, right):
    if left is None:
        return right
    if right is None:
        return left

    if left.data < right.data:
        left.next = merge(left.next, right)
        return left
    else:
        right.next = merge(left, right.next)
        return right


def merge_sort_ll(head):
    if head is None or head.next is None:
        return head

    slow = head
    fast = head

    # probably the most important thing here
    # this check will make slow the first element of the two
    # otherwise it will get stuck in a loop
    while fast.next and fast.next.next:
        slow = slow.next
        fast = fast.next.next

    mid = slow

    left = head
    right = mid.next
    mid.next = None

    left = merge_sort_ll(left)
    right = merge_sort_ll(right)

    return merge(left, right)


ll = take_input([45, 23, 121])
ll = merge_sort_ll(ll)
print_LL(ll)
```
