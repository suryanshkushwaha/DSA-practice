# Hashmap (hashtable)

- data structure storing key-value pair
- allows fast retrieval of values based on the keys by using a process called `hashing`
- each key is transformed into a unique index thorugh a hash function, which can be used to access the value (in constant time)

- student ID

# Practice Questions

```python
def count_freq(lst):
    hm = {}
    for item in lst:
        hm[item] = hm.get(item, 0) + 1

    return hm


nums = [1, 2, 2, 3, 3, 1, 1, 4, 2]
count_freq(nums)
```

```python
def group_anagrams(words):

    # thought process: why not sort the letters in the word
    # then each word who are anagrams will actually have the same seq of letter

    hm = {}

    for word in words:
        key = "".join(sorted(word))
        hm.setdefault(key, []).append(word)

    return hm


words = ["eat", "tea", "tan", "ate", "nat", "bat"]
group_anagrams(words)
```

```python
## 1. Intersection Of Two Arrays
def intersection_of_arrays(arr1, arr2):
    hm = {}

    for item in arr1:
        hm[item] = 1

    for item in arr2:
        hm[item] = hm.get(item, 0) + 1

    return [key for key, value in hm.items() if value > 1]


# Example Usage:
arr1 = [1, 2, 2, 1]
arr2 = [2, 2]
print(intersection_of_arrays(arr1, arr2))
```

```python
# 2. Check for Duplicates


def contains_duplicates(nums):
    hm = {}

    for item in nums:
        hm[item] = hm.get(item, 0) + 1
        if hm[item] > 1:
            return True

    return False


# Example Usage:
nums1 = [1, 2, 3, 1]
nums2 = [1, 2, 3, 4]
print(contains_duplicates(nums1))  # Output: True
print(contains_duplicates(nums2))  # Output: False
```

```python
# 3. First None Repeating Character in a string


def first_non_repeating_char(s):
    hm = {}

    for item in s:
        hm[item] = hm.get(item, 0) + 1

    for key, value in hm.items():
        if value == 1:
            return key

    return None


# Example Usage:
s = "swiss"
print(first_non_repeating_char(s))  # 'w'
```

# Implementing own HashMap: Hashing

```python
"""
in list retrieval is O(1)

if you got values 0,1,2,3,4
and u want to access let's say 3 directly without searching how can you do it?
just put them like this [0,1,2,3,4] now idx 3 is the value 3

therefore here hash(n) = n

but now let's say someone asked to store 0,1,2,3,4223994
if you create arry of such a big length like [0,1,2,3,........., 4223994] it's so space consuming. NOT Possible!

another problem
the `key` can be anything! it could be string as well like 'cheery', 'banana' etc
so what can we do?

create a hash func which converts the input key to NUMBERS!

we call this array we create as BUCKET ARRAY


another problem
what if the hash int received is greater than the size of bucket array, then what can you do?

HASH FUNCTION
1. hash code - key converted to integer
2. compress - key compressed to have an idx present in bucket array (like %)

what is a good hash function?
the one which is uniformly distributed, spread out, NOT concentreated

COLLISION HANDLING
if two hashes come out to be same even after compression.

    WAYS TO HANDLE
        1. Open Addressing (Closed hashing)
            - no of elems <= array list size
            - if initial place is empty save it else find an alternative
            - g(i) = [h(i) + k(i')] % size

            a. Linear probing k(i') -> i
                - g(i) = [h(i) + 0] % size
                - if idx is taken look right and if empty fill it, if not keep on looking right

            b. Quadratic Probing k(i') -> i^2
                - if idx is taken look right and if empty fill it, if not then look 2 places right, then 4 places then 8 then...

            c. Double hasing
                - f(i) = i * h'(i)   another hash
                - [h(i) + f(i)] % size

        2. Closed Addressing (Chaining)
            - will be saved at the SAME LOCATION
            - we have each pos of bucket as head of LL
            - it can be BST too!!
"""
```

```python
class HashMap:

    def __init__(self, capacity):
        self.capacity = capacity
        self.keys = [None] * self.capacity
        self.values = [None] * self.capacity
        self.size = 0

    def _hash_function(self, key):
        return abs(hash(key))

    def _rehash(self, old_hash_value):
        # linear probing to avoid collision
        return old_hash_value + 1

    def __resize(self):
        old_keys = self.keys
        old_values = self.values
        old_capacity = self.capacity

        self.capacity *= 2
        self.keys = [None] * self.capacity
        self.values = [None] * self.capacity
        self.size = 0

        for i in range(old_capacity):
            if old_keys[i] is not None:
                self.insert(old_keys[i], old_values[i])

    def insert(self, key, value):
        load_factor = self.size / self.capacity
        if load_factor >= 0.75:
            self.__resize()

        hash_value = self._hash_function(key)
        idx_to_input = hash_value % self.capacity
        probes = 0

        while probes < self.capacity:
            if self.keys[idx_to_input] is None or self.keys[idx_to_input] == "DELETED":
                self.keys[idx_to_input] = key
                self.values[idx_to_input] = value
                self.size += 1
                return

            if self.keys[idx_to_input] == key:
                self.values[idx_to_input] = value
                return

            hash_value = self._rehash(hash_value)
            idx_to_input = hash_value % self.capacity
            probes += 1

        return "HashMap is full; failed to insert"

    def get(self, key):
        hash_value = self._hash_function(key)
        initial_idx = hash_value % self.capacity
        probes = 0

        while probes < self.capacity:
            if self.keys[initial_idx] is None:
                return "Key Not Found"
            if self.keys[initial_idx] == key:
                return self.values[initial_idx]

            hash_value = self._rehash(hash_value)
            initial_idx = hash_value % self.capacity
            probes += 1

        return "Key Not Found"

    def delete(self, key):
        hash_value = self._hash_function(key)
        initial_idx = hash_value % self.capacity

        probes = 0

        while probes < self.capacity:
            if self.keys[initial_idx] is None:
                return "Key Not Found"
            if self.keys[initial_idx] == key:
                self.values[initial_idx] = "DELETED"
                self.keys[initial_idx] = None
                self.size -= 1
                return f"{key} deleted"

            hash_value = self._rehash(hash_value)
            initial_idx = hash_value % self.capacity
            probes += 1

        return "Key Not Found"

    def __len__(self):
        return self.size

    def __setitem__(self, key, value):
        return self.insert(key, value)

    def __getitem__(self, key):
        return self.get(key)

    def __str__(self):
        pairs = []
        for i in range(self.capacity):
            key = self.keys[i]
            if key is not None and key != "DELETED":
                pairs.append(f"{key}:{self.values[i]}")
        return "{" + ", ".join(pairs) + "}"
```

```python
hm = HashMap(8)

hm.insert(1, 10)
hm.insert(2, 20)
hm.insert(3, 40)

print(hm)
hm.insert(1, 10)
hm.insert(2, 20)
hm.insert(3, 100)
print(hm)

hm.insert(4, 10)
hm.insert(5, 20)
hm.insert(6, 100)
hm.insert(7, 10)
hm.insert(8, 20)
hm.insert(9, 100)
hm.insert(10, 100)
print(hm)

hm.delete(10)
hm.delete(8)
print(hm)
```

# HashMap: Chaining

```python
class LLNode:
    def __init__(self, key, value) -> None:
        self.key = key
        self.value = value
        self.next = None


class LinkedList:
    def __init__(self) -> None:
        self.head = None

    def add(self, key, value):
        newNode = LLNode(key, value)
        newNode.next = self.head
        self.head = newNode

    def search(self, key):
        curr = self.head

        while curr:
            if curr.key == key:
                return curr
            curr = curr.next

        return None

    def delete(self, key):
        if self.head is None:
            return False

        prev = self.head
        if self.head.key == key:
            self.head = self.head.next
            return True

        while prev.next:
            if prev.next.key == key:
                prev.next = prev.next.next
                return True
            prev = prev.next

        return False

    def traverse(self):
        curr = self.head

        while curr:
            print(f"{curr.key}:{curr.value} -->", end=" ")
            curr = curr.next

        print("None")


class HashMapChaining:

    def __init__(self, capacity=8) -> None:
        self.capacity = capacity
        self.size = 0
        self.buckets = self.__create_buckets(self.capacity)

    def __create_buckets(self, capacity):
        # buckets = [LinkedList()] * self.capacity --- This is wrong. THINK WHY!

        buckets = [LinkedList() for _ in range(self.capacity)]
        return buckets

    def __hash_function(self, key):
        return abs(hash(key))

    def __resize(self):
        old_buckets = self.buckets

        self.capacity *= 2
        self.size = 0

        self.buckets = self.__create_buckets(self.capacity)

        for item in old_buckets:
            curr = item.head
            while curr:
                self.insert(curr.key, curr.value)
                curr = curr.next

    def insert(self, key, value):
        load_factor = self.size / self.capacity
        if load_factor >= 0.75:
            self.__resize()

        hash_value = self.__hash_function(key)
        idx_to_insert = hash_value % self.capacity

        bucket = self.buckets[idx_to_insert]
        node = bucket.search(key)

        if node:
            node.value = value
            return f"{key} : {value} updated in HashMap"

        bucket.add(key, value)
        self.size += 1
        return f"{key} : {value} added in HashMap"

    def get(self, key):
        hash_value = self.__hash_function(key)
        idx_to_search = hash_value % self.capacity

        bucket = self.buckets[idx_to_search]
        node = bucket.search(key)

        if node:
            return node.value

        return "Key not found"

    def delete(self, key):
        hash_value = self.__hash_function(key)
        idx_to_delete = hash_value % self.capacity

        bucket = self.buckets[idx_to_delete]

        if bucket.delete(key):
            self.size -= 1
            return f"{key} deleted"

        return "Key not found"

    def __len__(self):
        return self.size

    def __setitem__(self, key, value):
        return self.insert(key, value)

    def __getitem__(self, key):
        return self.get(key)

    def __str__(self):
        for item in self.buckets:
            if item.head:
                item.traverse()

        return ""
```
