```python
"""
       node
       / \
    left  right

Each left and right node is a Binary Tree

IMPORTANT
        1            1
       /       ≠      \
      2                2
"""


class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


root = BinaryTreeNode(1)
root.left = BinaryTreeNode(2)
root.right = BinaryTreeNode(3)
```

```python
def print_BT(tree):
    if tree is None:
        return

    print(
        tree.data,
        ":",
        tree.left.data if tree.left else "None",
        ",",
        tree.right.data if tree.right else "None",
    )
    print_BT(tree.left)
    print_BT(tree.right)


print_BT(root)
```

```python
from collections import deque


def print_BT_levelwise(tree):
    if tree is None:
        return

    q = deque([tree])

    while q:
        tree = q.popleft()
        print(
            tree.data,
            ":",
            tree.left.data if tree.left else "None",
            ",",
            tree.right.data if tree.right else "None",
        )
        if tree.left:
            q.append(tree.left)
        if tree.right:
            q.append(tree.right)
```

```python
def take_input_BT():
    data = int(input("Enter data for Node"))
    if data == -1:
        return

    node = BinaryTreeNode(data)

    print(f"Enter left child of {data}")
    node.left = take_input_BT()

    print(f"Enter right child of {data}")
    node.right = take_input_BT()
```

```python
from collections import deque


def take_input_BT_levelwise():
    data = int(input("Enter data for Node"))

    root = BinaryTreeNode(data)
    q = deque([root])

    while q:
        node = q.popleft()
        left = int(input(f"Enter left child for parent node {node.data}"))
        right = int(input(f"Enter right child for parent node {node.data}"))
        if left != -1:
            node.left = BinaryTreeNode(left)
            q.append(node.left)
        if right != -1:
            node.right = BinaryTreeNode(right)
            q.append(node.right)

    return root


new_BT = take_input_BT_levelwise()
print_BT(new_BT)
```

```python
"""
max distance between two trees
upon looking closely, we can see that diameter = left height + right height
but then if a subtree is longer then our assumption is wrong

correct is max(diameter through root, left diameter, right diameter)
"""


def height(tree):
    if tree is None:
        return 0

    left_height = height(tree.left)
    right_height = height(tree.right)

    ans = 1 + max(left_height, right_height)

    return ans


def diameter(tree):
    if tree is None:
        return 0  # since tree diameter would be zero

    left_diameter = diameter(tree.left)
    right_diameter = diameter(tree.right)

    left_height = height(tree.left)
    right_height = height(tree.right)

    diameter_through_root = (
        left_height + right_height
    )  # very IMP NOT leftdiameter and rightdiameter

    return max(diameter_through_root, left_diameter, right_diameter)


# -- Test cases for diameter() on multiple tree shapes --
# 1) Balanced tree
root1 = BinaryTreeNode(1)
root1.left = BinaryTreeNode(2)
root1.right = BinaryTreeNode(3)
root1.left.left = BinaryTreeNode(4)
root1.left.right = BinaryTreeNode(5)
root1.right.left = BinaryTreeNode(6)
root1.right.right = BinaryTreeNode(7)

# 2) Skewed-left tree
root2 = BinaryTreeNode(1)
root2.left = BinaryTreeNode(2)
root2.left.left = BinaryTreeNode(3)
root2.left.left.left = BinaryTreeNode(4)

# 3) Skewed-right tree
root3 = BinaryTreeNode(1)
root3.right = BinaryTreeNode(2)
root3.right.right = BinaryTreeNode(3)
root3.right.right.right = BinaryTreeNode(4)

# 4) Unbalanced tree
root4 = BinaryTreeNode(1)
root4.left = BinaryTreeNode(2)
root4.right = BinaryTreeNode(3)
root4.left.left = BinaryTreeNode(4)
root4.left.left.left = BinaryTreeNode(5)
root4.right.right = BinaryTreeNode(6)
root4.right.right.right = BinaryTreeNode(7)

print("Diameter root1 (balanced):", diameter(root1))  # expected 4
print("Diameter root2 (skewed-left):", diameter(root2))  # expected 3
print("Diameter root3 (skewed-right):", diameter(root3))  # expected 3
print("Diameter root4 (unbalanced):", diameter(root4))  # expected 6
```

```python
"""
### Time Complexity in Tree

always think about
- complete trees
- linear trees

## Height of tree Complexity
O(n) n = no of nodes

## Diameter of tree Complexity
O(n.h)
- n = no of nodes
- h = height of tree

So for complete tree h = log n. Overall Complexity = O(n^h)
   for linear tree h = n. O(n^2)

"""
```

```python
"""
in the prev one,
for each subtree, height is called again and again
even tho for the parent node the height was called
and its again getting called on left and right

so why not
instead of just the height, we will ask the children to
get both height and diameter

"""


def diameter_optimized(tree):
    if tree is None:
        return 0, 0  # height, diameter

    left_height, left_diameter = diameter_optimized(tree.left)
    right_height, right_diameter = diameter_optimized(tree.right)

    diameter_through_root = left_height + right_height

    ans_diameter = max(diameter_through_root, left_diameter, right_diameter)
    curr_height = 1 + max(left_height, right_height)

    return curr_height, ans_diameter


treeHeight, treeDiameter = diameter_optimized(root1)
print(treeHeight, treeDiameter)

# NEW COMPLEXITY: O(n)
```

```python
"""
Given binary tree is balanced or not
if at each node,
    |left height - right height| <= 1    TREE IS HEIGHT BALANCED
"""


def is_balanced_BT(tree):
    if tree is None:
        return 0, True

    left_height, left_balanced = is_balanced_BT(tree.left)
    right_height, right_balanced = is_balanced_BT(tree.right)

    curr_tree_height = 1 + max(left_height, right_height)

    if not (left_balanced and right_balanced):
        return curr_tree_height, False

    if abs(left_height - right_height) <= 1:
        curr_tree_balanced = True
    else:
        curr_tree_balanced = False

    return curr_tree_height, curr_tree_balanced


ht, is_balanced = is_balanced_BT(root4)

print(ht, is_balanced)
```

```python
def preorderTraversal(root):
    if root is None:
        return

    print(root.data, end=" ")
    preorderTraversal(root.left)
    preorderTraversal(root.right)


preorderTraversal(root1)
```

```python
def postorderTraversal(root):
    if root is None:
        return

    postorderTraversal(root.left)
    postorderTraversal(root.right)
    print(root.data, end=" ")


postorderTraversal(root1)
```

```python
def inorderTraversal(root):
    if root is None:
        return

    inorderTraversal(root.left)
    print(root.data, end=" ")
    inorderTraversal(root.right)


inorderTraversal(root1)
```

```python
def create_tree_using_inorderAndPreorder(inorder, preorder):
    if not (inorder and preorder):
        return None

    root = preorder[0]
    rootNode = BinaryTreeNode(root)

    for idx, node in enumerate(inorder):
        if node == root:
            rootIdxInorder = idx
            break

    leftSubtreeInorder = inorder[:rootIdxInorder]
    rightSubtreeInorder = inorder[rootIdxInorder + 1 :]

    leftSubtreePreorder = preorder[1 : len(leftSubtreeInorder) + 1]
    rightSubtreePreorder = preorder[1 + len(leftSubtreeInorder) :]

    leftSubtree = create_tree_using_inorderAndPreorder(
        leftSubtreeInorder, leftSubtreePreorder
    )
    rightSubtree = create_tree_using_inorderAndPreorder(
        rightSubtreeInorder, rightSubtreePreorder
    )

    rootNode.left = leftSubtree
    rootNode.right = rightSubtree

    return rootNode


tree = create_tree_using_inorderAndPreorder([4, 2, 5, 1, 3, 6], [1, 2, 4, 5, 3, 6])
print_BT(tree)
```

```python
def create_tree_using_inorderAndPreorder_usingIdx(
    inorder, preorder, inS, inE, prS, prE
):
    if inS > inE or prS > prE:
        return None

    root = preorder[prS]
    rootNode = BinaryTreeNode(root)

    rootIdxInorder = inorder.index(root, inS, inE + 1)

    leftSubtree = create_tree_using_inorderAndPreorder_usingIdx(
        inorder, preorder, inS, rootIdxInorder - 1, prS + 1, rootIdxInorder - inS + prS
    )
    rightSubtree = create_tree_using_inorderAndPreorder_usingIdx(
        inorder, preorder, rootIdxInorder + 1, inE, rootIdxInorder - inS + prS + 1, prE
    )

    rootNode.left = leftSubtree
    rootNode.right = rightSubtree

    return rootNode


tree = create_tree_using_inorderAndPreorder_usingIdx(
    [4, 2, 5, 1, 3, 6], [1, 2, 4, 5, 3, 6], 0, 5, 0, 5
)
print_BT(tree)
```

```python
# LeetCODE 106
# # Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def buildTree(self, inorder, postorder):
        if not (inorder and postorder):
            return None

        root = postorder[-1]
        rootNode = BinaryTreeNode(root)

        rootNodeInorderIdx = inorder.index(root)

        leftSubtreeInorder = inorder[:rootNodeInorderIdx]
        rightSubtreeInorder = inorder[rootNodeInorderIdx + 1 :]

        leftSubtreePostorder = postorder[: len(leftSubtreeInorder)]
        rightSubtreePostorder = postorder[len(leftSubtreeInorder) : -1]

        rootNode.left = self.buildTree(leftSubtreeInorder, leftSubtreePostorder)
        rootNode.right = self.buildTree(rightSubtreeInorder, rightSubtreePostorder)

        return rootNode
```

## Types of Binary Trees

1. On the basis of no of children
    - **Full Binary Tree**
        - every parent/internal node has either 2 child OR no child

    - **Degenerate Binary Tree**
        - eevry internal node has JUST one child either left or right
        - analoous to linked list in terms of operation

    - **Skewed Binary Tree**
        - type of degen BT which is dominated by either the left node or the right node

2. On the basis of completion of level
    - **Complete Binary Tree**
        - all levels are completely filled except possibly the last level and the last level has the nodes as to the left as possible

    - **Perfect Binary Tree**
        - all internal node has 2 children and all leaf nodes are the same level
        - no of leaf nodes = 1 + internal nodes
        - no of nodes = 2^(n+1) - 1

    - **Balanced Binary Tree**
        - at every node height of left and right subtree for each node is either 0 or 1

3. Binary Search Tree (BST)
    - each node of tree follows 
        - left subtree contains node whose value is less than root's value
        - right subtree contains node whose value is greater than root's value
        - left and right subtree must also be BST
