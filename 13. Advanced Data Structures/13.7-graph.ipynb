# Graphs

- made up of set of nodes (called vertices) and edges (connection between pair of nodes)

## Real world examples
- Social Network 
- City Map and Roads
- Flight route and airports


## Industry use cases
- Google maps
- Search Engine
- Social Network Websites

Tree is a special kind of graph 
1. Every node is connected to each other in some way (No isolated node possible)
2. Trees dont have a cycle

## Terminologies

1. Node (vertex) - individual element in graph
2. Edge - connection between 2 vertices [unidirectional or bidirectional]
3. Adjacent Vertex - two vertices if they are connected by an edge
4. Degree - no of edges connected to node/vertex
    i. indegree - edges coming in 
    ii. outdegree - edges going out
5. Path - path of length n, from u to v is defined as seq of n+1 nodes p={u0, u1, ..., uN}
6. Cycle - if path start and end at the same vertex so u0 = uN

## Kinds of graph

1. Undirected Graph 
    - can go in either direction between nodes
2. Directed graph 
    - only go in the direction
3. Weighted graph / Unweighted
    - got some weight to the edge so edge got cost
4. Cyclic / Ayclic
    - contains one cycle atleast. 
5. Completed Graph 
    - all nodes are connected via edge to all the nodes
6. Labelled Graph 
    - any label to the edge between the nodes does not need to be numeric as in weighted graphs
7. Connected Graph
    - all nodes are connected via edge

```python
class GraphUsingEdgeList:
    def __init__(self) -> None:
        self.v = []
        self.edges = []

    def add_vertex(self, vertex):
        if vertex not in self.v:
            self.v.append(vertex)
        else:
            print(f"{vertex} already exists")

    def add_edge(self, source, destination, weight=1):
        if source in self.v and destination in self.v:
            self.edges.append((source, destination, weight))
        else:
            print("one or both vertex not found")

    def display(self):
        print("Vertices")
        if not self.v:
            print("None")
        else:
            for vertex in self.v:
                print(vertex, end=" ")
            print()

        print("Edges")
        for source, destination, _ in self.edges:
            print(f"{source} --> {destination}")


graph = GraphUsingEdgeList()
graph.add_vertex(1)
graph.add_vertex(2)
graph.add_vertex(3)

graph.add_edge(1, 2)
graph.add_edge(2, 3)


graph.display()

"""
Whats the problem with the above implementation?

think: how many edges in n node graph?
answer: in case of complete graph --> nC2 = n(n-1)/2 ~ n^2

THAT IS BAD

so now if you wanna lookup for an edge it will take n^2 time"""
```

```python
class GraphUsingAdjacencyList:
    def __init__(self) -> None:
        self.v = []
        self.adj_list = {}

    def add_vertex(self, vertex):
        if vertex not in self.v:
            self.v.append(vertex)
            self.adj_list[vertex] = []
        else:
            print(f"Vertex {vertex} already exists")

    def add_edges(self, source, destination, weight=1):
        if source in self.v and destination in self.v:
            self.adj_list[source].append((destination, weight))
            self.adj_list[destination].append((source, weight))
        else:
            print("one or both vertices not found")

    def display(self):
        for vertex, neighbour in self.adj_list.items():
            print(f"{vertex} : {neighbour}")


graph = GraphUsingAdjacencyList()
graph.add_vertex("A")
graph.add_vertex("B")
graph.add_vertex("C")
graph.add_vertex("D")
graph.add_vertex("E")

graph.add_edges("A", "B")
graph.add_edges("B", "E")


graph.display()

"""
O(n) lookup
this is because you will find the node in O(1)
but need to go through the list which will be O(n)
"""
```

```python
from collections import deque


class GraphUsingAdjacencyMatrix:
    def __init__(self, num_vertices) -> None:
        self.num_v = num_vertices
        self.v = [None] * self.num_v
        self.adj_matrix = [[0] * self.num_v for row in range(self.num_v)]

    def add_vertex(self, index, label):
        if index >= 0 and index < self.num_v:
            self.v[index] = label

        else:
            print("index OOB")

    def add_edge(self, source, destination, weight=1):
        if 0 <= source < self.num_v and 0 <= destination < self.num_v:
            self.adj_matrix[source][destination] = weight
            self.adj_matrix[destination][source] = weight

    def display(self):
        for index, label in enumerate(self.v):
            print(f"{index} - {label}")

        for row in self.adj_matrix:
            print(row)

    # Depth First Searchh
    def dfs(self, start_vertex=0):
        dfs_result = []
        visited = [False] * self.num_v
        self.dfs_recursive(start_vertex, dfs_result, visited)
        return dfs_result

    def dfs_recursive(self, vertex, dfs_result, visited):
        dfs_result.append(self.v[vertex])
        visited[vertex] = True

        for neighbour in range(self.num_v):
            if vertex == neighbour:
                continue

            if (self.adj_matrix[vertex][neighbour] >= 1) and (
                visited[neighbour] == False
            ):
                self.dfs_recursive(neighbour, dfs_result, visited)

    # Breadth First Search
    def bfs(self, start_vertex=0):
        bfs_result = []
        visited = [False] * self.num_v
        q = deque([start_vertex])

        while q:
            curr_node = q.popleft()
            visited[curr_node] = True
            bfs_result.append(self.v[curr_node])

            for neighbour in range(self.num_v):
                if curr_node == neighbour:
                    continue

                if (self.adj_matrix[curr_node][neighbour] >= 1) and (
                    visited[neighbour] == False
                ):
                    q.append(neighbour)
                    visited[neighbour] = True

        return bfs_result


graph = GraphUsingAdjacencyMatrix(6)
graph.add_vertex(0, "A")
graph.add_vertex(1, "B")
graph.add_vertex(2, "E")
graph.add_vertex(3, "D")
graph.add_vertex(4, "C")
graph.add_vertex(5, "F")

graph.add_edge(0, 1)
graph.add_edge(1, 3)
graph.add_edge(1, 2)
graph.add_edge(2, 3)
graph.add_edge(3, 4)
graph.add_edge(4, 5)

graph.display()

print("DFS")
print(graph.dfs())

print("BFS")
print(graph.bfs())
```

```python
"""
Traversal through disconnected graph
"""

from collections import deque


class GraphUsingAdjacencyMatrix:
    def __init__(self, num_vertices) -> None:
        self.num_v = num_vertices
        self.v = [None] * self.num_v
        self.adj_matrix = [[0] * self.num_v for row in range(self.num_v)]

    def add_vertex(self, index, label):
        if index >= 0 and index < self.num_v:
            self.v[index] = label

        else:
            print("index OOB")

    def add_edge(self, source, destination, weight=1):
        if 0 <= source < self.num_v and 0 <= destination < self.num_v:
            self.adj_matrix[source][destination] = weight
            self.adj_matrix[destination][source] = weight

    def display(self):
        for index, label in enumerate(self.v):
            print(f"{index} - {label}")

        for row in self.adj_matrix:
            print(row)

    # Depth First Searchh
    def dfs(self):
        dfs_result = []
        visited = [False] * self.num_v
        for vertex in range(
            self.num_v
        ):  # for disconnected graphs/ using this we can find the components/ using this we can find the path
            if not (visited[vertex]):
                self.dfs_recursive(vertex, dfs_result, visited)
        return dfs_result

    def dfs_recursive(self, vertex, dfs_result, visited):
        dfs_result.append(self.v[vertex])
        visited[vertex] = True

        for neighbour in range(self.num_v):
            if vertex == neighbour:
                continue

            if (self.adj_matrix[vertex][neighbour] >= 1) and (
                visited[neighbour] == False
            ):
                self.dfs_recursive(neighbour, dfs_result, visited)
```

```python
graph = GraphUsingAdjacencyMatrix(7)
graph.add_vertex(0, "A")
graph.add_vertex(1, "B")
graph.add_vertex(2, "E")
graph.add_vertex(3, "D")
graph.add_vertex(4, "C")
graph.add_vertex(5, "F")

graph.add_edge(0, 1)
graph.add_edge(1, 3)
graph.add_edge(1, 2)
graph.add_edge(2, 3)
graph.add_edge(3, 4)
graph.add_edge(4, 5)

graph.display()

print("DFS")
print(graph.dfs())
```
