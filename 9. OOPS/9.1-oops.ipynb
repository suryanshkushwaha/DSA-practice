```python
# instance variable and methods


class Dog:
    # constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age


## create object
dog1 = Dog("buddy", 3)
print(dog1.name)
print(dog1.age)
```

```python
# instance methods


class Dog:
    # constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof")


dog2 = Dog("no lah", 4)
dog2.bark()
```

```python
# model bank account


class BankAccount:
    def __init__(self, owner, balance=0) -> None:
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} is depositied. New bank balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("insufficient funds")
        else:
            self.balance -= amount
            print(f"{amount} withdrawn. New bank balance is {self.balance}")

    def get_balance(self):
        return self.balance


# create an account
account = BankAccount("suryansh", 5000)
account.balance
```

```python
account.deposit(120)
```

```python
account.withdraw(400)
```
