# Custom exception (Raise and Throw an exception)

```python
class Error(Exception):
    pass


class dobException(Error):
    pass
```

```python
year = int(input("enter the dob"))

age = 2026 - year

try:
    if age <= 20 and age >= 18:
        print("valid age, can apply for exams")
    else:
        raise dobException
except dobException:
    print("your age should be greater than 18 and less than 20")
```
