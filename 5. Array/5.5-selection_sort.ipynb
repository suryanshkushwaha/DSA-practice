# Selection Sort

- select the smallest element in each pass
- swap it to the right position
- after n pass, the first n positions will be correctly sorted

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(0, n):
        # pos=i
        # smallest=arr[i]
        min_idx = i
        for j in range(i, n):
            if arr[j] <= arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

    return arr


my_list = [
    3,
    234,
    6,
    45,
    234,
    65,
    835,
    234,
    624,
    64,
    3,
    24,
    5,
    2,
    245,
    534,
    700,
]

selection_sort(my_list)
```
