# insertion sort

- take the elems and place or insert them at the correct location
- each pass take in one extra elem then place it at the correct position
- to do so you will need to swap or shift the rest of the elems to the right to essentially insert it at the right location

```python
def insertion_sort(arr):
    n = len(arr)

    for i in range(n):    
        for j in range(i, 0,-1):
            if arr[j] < arr[j-1]:
                arr[j - 1], arr[j] = arr[j], arr[j-1]
            else:
                break
    return arr

my_list = [
    23,
    3,
    234,
    6,
    45,
    234,
    65,
    835,
    234,
    624,
    64,
    3,
    24,
    5,
    2,
    245,
    534,
    700,
]

insertion_sort(my_list)
```
