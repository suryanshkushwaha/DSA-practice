# Bubble Sort Algorithm

- largest element at any time is at the right location
- the elems next to each other are compared and swapped if necessary
- so essentially with each pass the the largest element will be placed at the correct location

```python
def bubble_sort(arr):

    for i in range(0, len(arr)):
        for j in range(0, len(arr) - i-1):
            if arr[j]>arr[j+1]:
                temp=arr[j]
                arr[j]=arr[j+1]
                arr[j+1]=temp

    return arr

my_list = [
    23,
    3,
    234,
    6,
    45,
    234,
    65,
    835,
    234,
    624,
    64,
    3,
    24,
    5,
    2,
    245,
    534,
    700,
]

bubble_sort(my_list)
```
