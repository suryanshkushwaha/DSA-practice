# searching and sorting

- linear search (arr, target)
- go through all the elems in loop and find it
- if cant find then return -1

```python
def linear_search(arr, target):
    for idx, i in enumerate(arr):
        if i==target:
            return idx
    return -1
    
my_list=[1,324,23,3,234,6,45,234,65,835,234,624,64,3,24,5,2,245,534,700]

linear_search(my_list, 700)
```
