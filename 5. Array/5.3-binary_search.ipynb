# binary search

- sorted array
- divides the array into 2

```python
def binary_search(arr, target):
    low=0
    high=len(arr)-1
    mid=0

    while low<=high:
        mid=(low+high)//2
        if arr[mid]==target:
            return mid
        elif arr[mid]<target:
            low=mid+1
        else: 
            high=mid-1

    return -1

my_list = [
    1,
    324,
    23,
    3,
    234,
    6,
    45,
    234,
    65,
    835,
    234,
    624,
    64,
    3,
    24,
    5,
    2,
    245,
    534,
    700,
]
my_list.sort()
print(my_list)
binary_search(my_list, 700)
```
