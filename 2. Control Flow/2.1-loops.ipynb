```python
# for loop

for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-1):
    print((i))
```

```python
# strings

str='suryansh kushwaha'

for i in str:
    print(i)
```

```python
# while loop

count=0

while count<5:
    print(count)
    count=count+1
```

```python
count =0

while count%2==0:
    print(count)
    count=count+1
```

```python
# loop control statements

## break
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass - does nothing

for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
# nested loops

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
# example - prime no between 1 & 100

for num in range(1, 101):
    for i in range(2, num):
        if num%i==0:
            break
    else: # used with a for loop, will run compulsory at the end of the loop
        print(num)
```
