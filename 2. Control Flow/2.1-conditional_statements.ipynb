```python
# if statement
age = 23

if age >= 18:
    print("can vote")
```

```python
# else statement
age = 16
if age >= 18:
    print("can vote")
else:
    print("you are minor")
```

```python
# elif statement

age = 14

if age < 13:
    print("you are a kid")
elif age < 18:
    print("you are a teen")
else:
    print("an adult now")
```

```python
# nested conditional statement

num = int(input("enter a no"))

if num >= 0:
    print("positive no")
    if num % 2 == 0:
        print("even no")
    else:
        print("odd no")
else:
    print("zero or negative")
```
