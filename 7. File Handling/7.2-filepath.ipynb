```python
import os

cwd = os.getcwd()
print(f"current working directory is {cwd}")
```

```python
new_dir = "package"
os.mkdir(new_dir)
print(f"dicrectory {new_dir} created")
```

```python
# list files and directories

items = os.listdir(".")
print(items)
```

```python
# joining paths

dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd(), dir_name, file_name)
print(full_path)
```

```python
path = "example1.txt"

if os.path.exists(path):
    print("exists")
else:
    print("not here")
```

```python
# check path is file or dir

path = "example.txt"

if os.path.isfile(path):
    print("file")
elif os.path.isdir(path):
    print("dir")
else:
    print("neither")
```

```python
# getting the absolute path

relative_path = "example.bin"
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
