# File Operation - Read and Write Files

```python
# Read a whole file

with open("example.txt", "r") as file:
    print(file.read())
```

```python
# read file line by line
with open("example.txt", "r") as file:
    for line in file:
        print(line)
```

```python
with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())  # removes the newline char
```

```python
# write a file (overwriting)

with open("example.txt", "w") as file:
    file.write("i am a new content muahahah")
```

```python
# write a file (without overwriting)
with open("example.txt", "a") as file:
    file.write("now this will be appended because of append mode")
```

```python
# write list of lines to a file
lines = ["file line \n", "second line \n", "third line \n"]
with open("example.txt", "a") as file:
    file.writelines(lines)
```

```python
# BINARY FILE

data = b"\x00\x11"
with open("example.bin", "wb") as file:
    file.write(data)
```

```python
with open("example.bin", "rb") as file:
    content = file.read()
    print(content)
```

```python
# Read a text file and count the no of lines, words and characters

words = []
with open("example.txt", "r") as file:
    line_c = 0
    for line in file:
        line_c += 1
        words.extend(line.split())
    words_c = len(words)
    char_c = 0
    for word in words:
        char_c += len(word)

print(line_c, words_c, char_c)
```

```python
# writing and then reading a file

with open("example.txt", "w+") as file:
    file.write("hello word \n")
    file.write("this is another line")

    # move the cursor to the beginnning
    # file.seek(0) # UNCOMMENT THIS TO SEE THE CHANGE

    content = file.read()
    print(content)
```
