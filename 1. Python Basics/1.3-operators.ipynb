```python
a = 10
b = 5

add = a + b
sub = a - b
mul = a * b
div = a / b
floor_div = a // b
mod = a % b
exp = a**b

print(add)
print(sub)
print(mul)
print(div)
print(floor_div)
print(mod)
print(exp)
```

```python
# Logical operators

## and, or, not
X = True
Y = False

print(X and Y)
print(X or Y)
print(not X)
```
