```python
# variable types
# dynamic typed language - type determined at run time

age = 23
print(type(age))

# type conversion
age_str = str(age)
print(type(age_str))
```

```python
age = "25"
int(age)  # this wont change the type but you can convert it here
age_int = int(age)  # this will assign the int value of str to age_int

print(type(age))
print(type(age_int))
```

```python
# dynamic typing

var = 10
print(var, type(var))

var = "hello"
print(var, type(var))
```
