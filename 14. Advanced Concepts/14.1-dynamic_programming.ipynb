# Fibonacci Series

1 1 2 3 5 8 13 ...

we are calculating the result of each n on each call
Time complexity = 2^n

```python
calls = {}


def fibonacci(n):
    if n not in calls:
        calls[n] = 1
    else:
        calls[n] += 1

    if n <= 1:
        return 1

    fib1 = fibonacci(n - 1)
    fib2 = fibonacci(n - 2)

    return fib1 + fib2


print(fibonacci(25))
print(calls)
```

```python
# Memoization

memo = [-1] * 6


def fib_dp(n):
    if memo[n] != -1:
        return memo[n]

    if n <= 1:
        return 1

    fib1 = fib_dp(n - 1)
    fib2 = fib_dp(n - 2)

    memo[n - 1] = fib1
    memo[n - 2] = fib2
    memo[n] = fib1 + fib2

    return memo[n]


fib_dp(5)

# now the complexity is O(n)
```

```python
# bottom-up


def fib_bottom_up(n):
    if n <= 1:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]


fib_bottom_up(5)

# Time Complexity: O(n)
```

# Minimum steps to 1

-- Given number n, we have to tell min number of steps required to reduce the no to 1. Below operations are allowed
1. Subtract 1
2. Divide by 2
3. Divide by 3

```python
def min_steps(n):
    if n <= 1:
        return 0

    sub1 = min_steps(n - 1)
    if n % 2 == 0:
        div2 = min_steps(n // 2)
    else:
        div2 = float("inf")
    if n % 3 == 0:
        div3 = min_steps(n // 3)
    else:
        div3 = float("inf")

    ans = min(sub1, div2, div3) + 1

    return ans


min_steps(7)
```

```python
# memo
def min_steps(n, memo):
    if n <= 1:
        memo[n] = 0
        return 0
    if memo[n] != -1:
        return memo[n]

    sub1 = min_steps(n - 1, memo)
    if n % 2 == 0:
        div2 = min_steps(n // 2, memo)
    else:
        div2 = float("inf")
    if n % 3 == 0:
        div3 = min_steps(n // 3, memo)
    else:
        div3 = float("inf")

    ans = min(sub1, div2, div3) + 1
    memo[n] = ans

    return ans


min_steps(7, [-1] * 8)
```

```python
# bottom up


def min_steps(n):
    dp = [-1] * (n + 1)
    dp[1] = 0

    for i in range(2, n + 1):
        sub = dp[i - 1]
        div2 = dp[i // 2] if i % 2 == 0 else float("inf")
        div3 = dp[i // 3] if i % 3 == 0 else float("inf")
        ans = min(sub, div2, div3) + 1
        dp[i] = ans

    return dp[n]


min_steps(7)
```

```python
# faster way of writing this
def min_steps(n):
    dp = [-1] * (n + 1)
    dp[1] = 0

    for i in range(2, n + 1):
        ans = dp[i - 1] + 1
        if i % 2 == 0:
            ans = min(ans, dp[i // 2] + 1)
        if i % 3 == 0:
            ans = min(ans, dp[i // 3] + 1)

        dp[i] = ans

    return dp[n]


min_steps(7)
```

```python
# WAP to find the total no of possible balanced trees for a given h
# for a tree to be balanced the |left subtree - right subtree| <= 1


def count_balanced_tree(h):
    if h == 0 or h == 1:
        return 1  # for height 0, i can say that tree with node none is present

    h1 = count_balanced_tree(h - 1)
    h2 = count_balanced_tree(h - 2)

    # ans = if both h1 + (left h1, right h2) + (left h2, right h1)
    ans = h1 * h1 + h1 * h2 + h2 * h1

    # why multiplication because we need to find total no of ways
    # take for example two dice, how many possible combinations are possible
    # 6*6 = 36

    return ans


count_balanced_tree(4)
```

```python
"""
You have a cost matrix called cost[][], and you need to find the minimum cost 
to reach a specific position (M, N) starting from the top-left corner (0, 0).

Each cell in the matrix has a cost that you need to add up to get the total cost 
for the path you take. You can only move down, to the right, or diagonally 
down-right from any cell. 

Keep in mind that all costs are positive integers.
"""


def min_cost_path_recursive(cost, m, n):
    if m == 0 and n == 0:
        return cost[0][0]

    if m < 0 or n < 0:
        return float("inf")

    up = min_cost_path_recursive(cost, m - 1, n)
    left = min_cost_path_recursive(cost, m, n - 1)
    diag = min_cost_path_recursive(cost, m - 1, n - 1)

    ans = cost[m][n] + min(up, left, diag)

    return ans


cost_matrix = [[1, 2, 3], [4, 8, 2], [1, 5, 3]]
M, N = 2, 2
min_cost_path_recursive(cost_matrix, M, N)
```

```python
def min_cost_path_memo(cost, m, n):
    if m == 0 and n == 0:
        return cost[0][0]

    if m < 0 or n < 0:
        return float("inf")

    if memo[m][n] != -1:
        return memo[m][n]

    up = min_cost_path_memo(cost, m - 1, n)
    left = min_cost_path_memo(cost, m, n - 1)
    diag = min_cost_path_memo(cost, m - 1, n - 1)

    ans = cost[m][n] + min(up, left, diag)
    memo[m][n] = ans
    return ans


cost_matrix = [[1, 2, 3], [4, 18, 12], [1, 5, 3]]
M, N = 2, 2
memo = [[-1 for i in range(N + 1)] for j in range(M + 1)]
min_cost_path_memo(cost_matrix, M, N)
```

```python
def min_cost_path_tabulation(cost, m, n, dp):

    # # base case
    # dp[0][0] = cost[0][0]

    for row in range(m + 1):
        for col in range(n + 1):
            if row == 0 and col == 0:
                dp[row][col] = cost[0][0]  # base case
            else:
                cost_left = dp[row][col - 1] if col > 0 else float("inf")
                cost_top = dp[row - 1][col] if row > 0 else float("inf")
                cost_diag = (
                    dp[row - 1][col - 1] if (row > 0 and col > 0) else float("inf")
                )

                dp[row][col] = cost[row][col] + min(cost_diag, cost_top, cost_left)

    return dp[m][n]


cost_matrix = [[1, 2, 3], [4, 18, 12], [1, 5, 3]]
M, N = 2, 2
dp = [[0 for i in range(N + 1)] for j in range(M + 1)]
min_cost_path_tabulation(cost_matrix, M, N, dp)
```
