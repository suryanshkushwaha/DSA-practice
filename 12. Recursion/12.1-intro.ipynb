# Recursion
It is when soln of problem depends on same smaller problem


- call a func from inside of another func
- func stay in memory until it gets fully executed

- doing almost similar work in all the func

```python
def print_num(num):
    print(num)
    if num == 1:
        return
    print_num(num - 1)


print_num(5)
```

```python
import sys

sys.getrecursionlimit()
```

```python
def fact(n):
    return n * fact(n - 1) if n > 1 else 1


fact(5)
```

```python
def power2(n):
    return 2 * power2(n - 1) if n > 1 else 2


power2(5)
```
