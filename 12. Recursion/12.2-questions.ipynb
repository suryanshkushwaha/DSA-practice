## Q1. Find sum of all natural numbers upto a given number (n)

### State the recurrence relation and draw the recursion tree

```python
"""
sum(n) = n + (n-1) + (n-2) + ... + 1
sum(n) = n + sum(n-1)
"""


def sum_of_natural_number(n):
    return n + sum_of_natural_number(n - 1) if n > 1 else 1


sum_of_natural_number(5)
```

## Q2. Find no of digits in a number

```python
"""
let num = 23476
num_of_digits = 1 + num_of_digits(num//10)
"""


def num_of_digits(n):
    if n >= 1 and n <= 9:
        return 1
    elif n == 0:
        return 1
    else:
        return 1 + num_of_digits(n // 10)


num_of_digits(224323)
```

## Q3. Fibonacci Series

1 1 2 3 5 8 13 ...

```python
"""
f(n) = f(n-1) + f(n-2)
"""


def f(n):
    if n == 0 or n == 1:
        return 1
    else:
        return f(n - 1) + f(n - 2)


f(6)
```

## Q4. For a given no n

### 1. print 1 to N
### 2. print N to 1

```python
# 1


def one_to_n(n):
    if n == 0:
        return
    one_to_n(n - 1)  # HEAD RECURSION
    print(n)


one_to_n(5)
```

```python
# 2


def n_to_one(n):
    if n == 0:
        return
    print(n)
    n_to_one(n - 1)  # TAIL RECURSION


n_to_one(5)
```

## Q5. sum of digit of a number

```python
def sum_of_digit(n):
    return n % 10 + sum_of_digit(n // 10) if n >= 1 else 0


sum_of_digit(12455234)
```

## Q6. power of a number (base, exp)

```python
def power(base, exp):
    return base * power(base, exp - 1) if exp > 0 else 1


power(5, 5)
```
