## Merge Sort using Recursion

```python
def merge(left, right):
    if len(left) == 0:
        return right
    if len(right) == 0:
        return left

    if left[0] <= right[0]:
        return [left[0]] + merge(left[1:], right)

    return [right[0]] + merge(left, right[1:])


def merge_sort(lst):
    if len(lst) <= 1:
        return lst

    mid = (len(lst)) // 2
    left = lst[:mid]
    right = lst[mid:]

    left = merge_sort(left)
    right = merge_sort(right)

    return merge(left, right)


print(merge_sort(lst))
```

```python
# save space - inplace


def merge(lst, i, s, m, j, e):
    if i > m:
        return lst[j : e + 1]
    if j > e:
        return lst[i : m + 1]
    if lst[i] <= lst[j]:
        return [lst[i]] + merge(lst, i + 1, s, m, j, e)
    return [lst[j]] + merge(lst, i, s, m, j + 1, e)


def merge_sort(lst, s, e):
    if s >= e:
        return
    m = s + (e - s) // 2
    merge_sort(lst, s, m)
    merge_sort(lst, m + 1, e)
    merged = merge(lst, s, s, m, m + 1, e)
    lst[s : e + 1] = merged


merge_sort(lst, 0, len(lst) - 1)
lst
```

## Quick Sort using Recursion

unlike merge sort, we first do work then let recursion handle the rest

```python
def partition(lst, s, e):
    pivot = lst[e]
    i = s
    for j in range(s, e):
        if lst[j] <= pivot:
            lst[i], lst[j] = lst[j], lst[i]
            i += 1
    lst[i], lst[e] = lst[e], lst[i]
    return i


def quick_sort(lst, s, e):
    if s >= e:
        return
    pivot_idx = partition(lst, s, e)
    quick_sort(lst, s, pivot_idx - 1)
    quick_sort(lst, pivot_idx + 1, e)


quick_sort(lst, 0, len(lst) - 1)
```
