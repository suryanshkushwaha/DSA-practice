| Feature                | Head Recursion                                               | Tail Recursion                                               |
|:----------------------:|:-----------------------------------------------------------:|:------------------------------------------------------------:|
| **Working**            | 1. Base Case<br>2. Recursion<br>3. Work                     | 1. Base Case<br>2. Work<br>3. Recursion                     |
| **Description**        | Recursive call at the beginning of function implementation  | Recursive call at the end of function implementation         |
| **Python Support**     | Supported by default                                        | Not optimized by default                                     |
| **Memory Usage**       | Higher (for deep recursion)                                 | Lower (more efficient for deep recursion)                    |
| **Stack Usage**        | Function calls stored in stack memory                       | Function calls not stored in stack memory (if optimized)     |

```python
def fac_head(n):
    if n == 1:
        return 1
    small_ans = fac_head(n - 1)
    final_ans = n * small_ans
    return final_ans


def fac_tail_WRONG(n):
    if n == 1:
        return 1

    return n * fac_tail_WRONG(n - 1)


# the above example is to show that even it seems like the func is called at the end, but it in reality is NOT.
# it is doing the work AFTER the function call making it a head recursion


def fac_tail(n, accumulator=1):
    if n == 1:
        return accumulator
    accumulator *= n
    return fac_tail(n - 1, accumulator)
```
