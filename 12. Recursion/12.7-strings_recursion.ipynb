## Strings using recursion

string = 'abc'


**substring** = ordering to be maintained and it should be continuous

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;= a  b  c

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;ab bc abc

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;= n(n+1)/2

**subsequence** = order should be maintained and it can be NON-continuous

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;= a   b   c

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;ab  bc  ac

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;abc φ

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;= 2^n

**permutation** = all possible arrangements of the string (order matters, all characters used)

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;abc, acb, bac, bca, cab, cba

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;= n!

```python
def remove_character(str, ch):
    if len(str) == 0:
        return ""
    if str[0] != ch:
        return str[0] + remove_character(str[1:], ch)
    else:
        return remove_character(str[1:], ch)


remove_character("good morningzzz", "z")
```

```python
str = '"nitin"'


def is_palindrome(str):
    if len(str) <= 1:
        return True
    if str[0] == str[-1]:
        return is_palindrome(str[1:-1])
    else:
        return False


is_palindrome(str)
```

```python
def return_subsequences(str):
    if len(str) == 0:
        return [""]
    small_ans = return_subsequences(str[1:])

    ans = []
    ans.extend(small_ans)

    for i in small_ans:
        ans.append(str[0] + i)

    return ans


return_subsequences("abc")
```

```python
def print_subsequences(str, takenSoFar):
    if len(str) == 0:
        print(takenSoFar)
        return
    curr_char = str[0]
    small_input = str[1:]

    print_subsequences(small_input, takenSoFar + curr_char)
    print_subsequences(small_input, takenSoFar)

    return


print_subsequences("abc", "")
```

```python
def print_permutations(str, takenSoFar):
    if str == "":
        print(takenSoFar)
        return

    for i in range(0, len(takenSoFar) + 1):
        print_permutations(str[1:], takenSoFar[0:i] + str[0] + takenSoFar[i:])

    return


print_permutations("abc", "")
```

```python
def return_permutations(str):
    if str == "":
        return [""]

    small_ans = return_permutations(str[1:])

    curr_char = str[0]
    ans = []

    for item in small_ans:
        for i in range(0, len(item) + 1):
            ans.append(item[0:i] + curr_char + item[i:])

    return ans


return_permutations("abc")
```

```python
# KeyPad Words
numpad_dict = {
    1: [],
    2: ["a", "b", "c"],
    3: ["d", "e", "f"],
    4: ["g", "h", "i"],
    5: ["j", "k", "l"],
    6: ["m", "n", "o"],
    7: ["p", "q", "r", "s"],
    8: ["t", "u", "v"],
    9: ["w", "x", "y", "z"],
    0: [],
}


def return_words(str):
    if str == "":
        return [""]

    small_ans = return_words(
        str[1:]
    )  # i wish they return me all the permutations of the rest of the numbers

    ans = []
    curr_digit = str[0]
    alphabets = numpad_dict[int(curr_digit)]

    for ch in alphabets:
        ans.extend(map(lambda x: ch + x, small_ans))

    return ans


return_words("3825968")
```

```python
alphabet_dict = {i: chr(96 + i) for i in range(1, 27)}


def return_all_codes(str):
    if str == "":
        return [""]

    if len(str) == 1:
        return [alphabet_dict[int(str)]]

    ans = []
    first = int(str[0])
    first_two = int(str[0:2])

    if first <= 26:
        small_ans = return_all_codes(str[1:])
        for i in small_ans:
            ans.append(alphabet_dict[first] + i)
    if first_two <= 26 and first_two >= 10:
        small_ans_two = return_all_codes(str[2:])
        for i in small_ans_two:
            ans.append(alphabet_dict[first_two] + i)

    return ans


return_all_codes("678912")
```

```python
def tower_of_hanoi(n, s, d, a):
    if n == 1:
        print(s, "->", d)
        return
    if n == 0:
        return

    tower_of_hanoi(n - 1, s, a, d)
    print(s, "-->", d)
    tower_of_hanoi(n - 1, a, d, s)


tower_of_hanoi(3, "s", "d", "a")
```
