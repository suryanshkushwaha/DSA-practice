# filter()

- filter out based on condition

```python
# syntax

filter( func , lst)
```

```python
nums = [1,2,3,4,5,6,7,8]

list(filter(lambda x:x%2==0, nums))
```

```python
list(filter(lambda x:x>5 and x%2==0, nums))
```

```python
people = [{"name": "suryansh", "age": 23}, {"name": "jack", "age": 33}]

age_greater_than_25 = lambda person:person['age']>25

list(filter(age_greater_than_25, people))
```
