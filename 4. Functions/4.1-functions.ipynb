# Functions

- block of code
- performs specific task
- organize code, reuse, readability

```python
# syntax


def function_name(parameters):
    """Docstring"""
    # function body
    return expression
```

```python
# create func


def even_or_odd(n):
    if n % 2 == 0:
        print("even")
    print("odd")
```

```python
# call func

even_or_odd(5)
```

```python
# multiple params


def add(a, b):
    return a + b
```

```python
# default params


def greet(name="guest"):
    print(f"Hello {name}")


greet("suryansh")
```

```python
greet()
```

```python
# variable length arguments

## positional and keyword arguments


# def print_nums(num1, num2, num3, num4, num5):
def print_nums(*args):
    for nums in args:
        print(nums)
```

```python
print_nums(1, 2, 3, 4, 5, 6, 7, "suryansh")
```

```python
## keyword argument


def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="krish", age=32, country="singapore")
```

```python
# you can even do both together, it will understand which is positional and which is keyword
```

```python
# return - can return multiple params


def multiply(a, b):
    return a, b, a * b
```

```python
multiply(5, 4)
```
