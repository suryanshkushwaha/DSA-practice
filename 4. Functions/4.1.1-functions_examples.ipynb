Example1. Temperature Conversion

```python
def convert_temp(temp, unit):
    """Convert C to F or F to C"""
    match unit:
        case 'F':
            return str(float((temp - 32) * 5/9))+'C'
        case 'C':
            return str(float((temp * 9/5) + 32))+'F'
        
print(convert_temp(37,'C'))
```

Example2. Password Strength Checker

```python
def is_strong_password(pwd):
    if len(pwd)<=10:
        return False
    if not any(char.isdigit() for char in pwd):
        return False
    if not any(char.islower() for char in pwd):
        return False
    if not any(char.isupper() for char in pwd):
        return False
    if not any(char in '!@#$%^&*()_+' for char in pwd):
        return False
    return True

print(is_strong_password('abrak3aEasdfa!dsf'))
```

Example3. Total Cost of items in Shopping Cart

```python
def total_cost(cart):
    total_cost=0
    product_price=0
    for product in cart:
        product_price = product["qty"] * product["price"]
        total_cost+=product_price
        print(f"{product['item']} : {product['qty']} : {product_price}")
        product_price=0
    return total_cost

cart = [
    {'item':'Apple', 'qty': 4, 'price':2},
    {'item':'Banana', 'qty': 12, 'price':0.5},
    {'item':'Maggi Noodles', 'qty': 3, 'price':3}
]
print(total_cost(cart))
```

Example4. Palindrome String

```python
def is_palindrome(str):
    str=str.lower().strip().replace(' ','')
    return str==str[::-1]
    
print(is_palindrome('aAAa  aa'))
```

Example5. Factorial Using Recursion

```python
def factorial(n):
    if n<=1:
        return n
    return n*factorial(n-1)

print(factorial(5))
```

Example6. Read a file and print freq of the words init.

```python
def word_freq(file_path):
    words=[]
    word_freq_dict={}
    with open(file_path, 'r') as file:
        for line in file:
            words.extend(line.split())
    for word in words:
        word=word.lower().strip('.,!\'\"@#$%^&*()_+=`~')
        word_freq_dict[word]=word_freq_dict.get(word, 0)+1
    
    return word_freq_dict

print(word_freq('./4.1.1-sample.txt'))
```

Example7. Validate email address

```python
import re
def is_valid_email(mail):
    pattern = r'^[a-zA-Z0-9_.-]+@[a-zA-Z0-9_-]+\.[a-zA-Z0-9-.]+$'
    return re.match(pattern, mail) is not None

print(is_valid_email('suryanshkushwaha@u.nus.edu'))
```
