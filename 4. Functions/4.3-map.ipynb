```python
# map() - applies func to all items in list

nums=[1,2,3,4,5,6]

map(lambda x:x**2, nums) # lazy loading
```

```python
list(map(lambda x:x**2, nums))
```

```python
# map multiple iterables
nums1=[1,2,3]
nums2=[4,5,6]
add = lambda x,y:x+y

list(map(add, nums1, nums2))
```

```python
# map() to convert list of strings to integers
str_nums=['1','2','3','4']
int_nums=list(map(int, str_nums))

print(int_nums)
```

```python
words=['apple', 'banana', 'cherry']

uppercase_words=list(map(str.upper, words))
print(uppercase_words)
```

```python
def get_name(person):
    return person['name']

people = [
    {'name':'suryansh', 'age':23},
    {'name':'jack', 'age': 33}
]

list(map(get_name, people))
```
