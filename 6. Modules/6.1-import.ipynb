modules and packages help organize and reuse code. 

```python
import math

math.sqrt(16)
```

```python
from math import sqrt, pi

sqrt(16)
print(pi)
```

```python
import numpy as np

np.array([1, 2, 3, 4])
```

```python
from math import *  # imports all the packages

cos(54)
```

```python
from package.maths import add

add(2, 3)
```

```python
from package.maths import *
from package.subpackages.mult import mult

print(add(2, 3))
print(sub(3, 2))

print(mult(2, 3))
```
